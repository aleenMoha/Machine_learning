### Project Idea: Predict Students Pass/Fail Depends On Multiple Feature
Logistic Regression

```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
# DATASET
# Features: [Hours Studied, Attendance Rate (%), IQ Score]
x = np.array( [
    [1, 70, 95],
    [2, 75, 100],
    [3, 80, 102],
    [4, 85, 105],
    [5, 90, 110],
    [6, 92, 108],
    [7, 95, 115],
    [8, 98, 120]
])
y = np.array([0, 0, 0, 1, 1, 1, 1, 1]) #pass = 1 /fail = 0
```

```python
#IMPORTANT FUNCTIONS

## Z-SCORE NORMALIZATION
def zscor (x):
    mu = np.mean(x,axis=0)
    sig = np.std(x,axis=0)
    xnorm = (x-mu)/sig
    return xnorm,mu,sig

## SIGMOID FUNCTION
def sigmoid (z):
    g = 1 / (1+np.exp(-z))
    return g

## COST OF AN EXAMPLE
def cost (x,y,w,b, lmbd):
    m,n = x.shape
    c = 0
    for i in range(m):
        z = np.dot(x[i],w) + b
        f = sigmoid(z)
        c += - y[i] * np.log(f) - (1-y[i]) * np.log(1-f)
    t = c / m
    reg = 0
    for j in range(n):
        reg += (w[j]**2)
    reg *= lmbd/(2*m)
    t += reg
    return t

#COMPUTE GRADIENT OF MULTIPLE FEATURES
def grad (x,y,w,b,  lmbd):
    m,n = x.shape
    dw = np.zeros(n)
    db =0.
    for i in range (m):
        err = sigmoid(np.dot(x[i],w) + b) - y[i]
        for j in range (n):
            dw[j] += err*x[i,j]
        db += err
    dw /= m
    db /= m
    for j in range(n):
       dw[j] += (lmbd/m) * w[j]
    return dw,db

# GRADIENT DESCENT
def desc (x,y,w,b,cost,grad,alpha,num, lmbd):
    j=[]
    for i in range (num):
        dw,db = grad (x,y,w,b, lmbd)
        w -= alpha *dw
        b -= alpha *db

        if i <100000:
            j.append(cost(x,y,w,b, lmbd))
        if i % math.ceil(num/10) == 10 :
            print(f'iterations:{i:4}, cost:{j[-1]:0.3e}')
    return w,b,j
```

```python
#PREDICT VALUES
def compf (x,w,b):
    f = sigmoid(np.dot(x, w) + b)
    if f<=0.5:
        f=0
    else:
        f=1
    return f
```

```python
xnorm,mu,sig = zscor(x)
w,b,j = desc(xnorm,y,np.array([0.,0.,0.]), 0., cost, grad, 0.00001, 1000, 0.7)
print(w, b)
```

```python
xt = np.array(  [4, 70, 130])
xnorm = (xt - mu) / sig
f = compf(xnorm, w,b)
if f == 0:
    print('fail')
else:
    print('pass')
print(f)
```
